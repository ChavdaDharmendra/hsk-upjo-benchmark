```python
# Features to be used for prediction
features = [
    "Age_years",
    "Sex",
    "Laterality",
    "Prenatal_hydronephrosis",
    "Initial_presentation",
    "Flank_pain",
    "Recurrent_UTI",
    "Hydronephrosis_SFU_grade",
    "APD_mm",
    "Preoperative_DRF_percent",
    "Diuretic_renogram_T1_2_min",
    "High_ureteral_insertion",
    "Crossing_vessel",
    "Renal_stone"
]

# Input features
X = df[features].copy()

# Target variable
y = df["UPJO"].copy()

print("X Shape:", X.shape)
print("y Shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget distribution:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

# Identify categorical and numerical features
categorical_features = [
    "Sex",
    "Laterality",
    "Prenatal_hydronephrosis",
    "Initial_presentation",
    "Flank_pain",
    "Recurrent_UTI",
    "High_ureteral_insertion",
    "Crossing_vessel",
    "Renal_stone"
]

numerical_features = [
    "Age_years",
    "Hydronephrosis_SFU_grade",
    "APD_mm",
    "Preoperative_DRF_percent",
    "Diuretic_renogram_T1_2_min"
]

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                drop="first"
            ),
            categorical_features
        )
    ],
    remainder="passthrough"
)

# Fit preprocessing only on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform test data using the already-fitted preprocessor
X_test_processed = preprocessor.transform(X_test)

print("\nProcessed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# Create Logistic Regression model
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Train the model
logistic_model.fit(X_train_processed, y_train)

# Predictions
y_pred_lr = logistic_model.predict(X_test_processed)

# Probability predictions for ROC-AUC
y_prob_lr = logistic_model.predict_proba(X_test_processed)[:, 1]

# Evaluation metrics
accuracy_lr = accuracy_score(y_test, y_pred_lr)
precision_lr = precision_score(y_test, y_pred_lr)
recall_lr = recall_score(y_test, y_pred_lr)
f1_lr = f1_score(y_test, y_pred_lr)
roc_auc_lr = roc_auc_score(y_test, y_prob_lr)

# Confusion Matrix
cm_lr = confusion_matrix(y_test, y_pred_lr)

print("===== Logistic Regression Results =====")
print("Accuracy :", accuracy_lr)
print("Precision:", precision_lr)
print("Recall   :", recall_lr)
print("F1 Score :", f1_lr)
print("ROC-AUC  :", roc_auc_lr)

print("\nConfusion Matrix:")
print(cm_lr)

print("\nClassification Report:")
print(classification_report(y_test, y_pred_lr))
```

```python
from sklearn.ensemble import RandomForestClassifier

# Create Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    class_weight="balanced"
)

# Train
rf_model.fit(X_train_processed, y_train)

# Predictions
y_pred_rf = rf_model.predict(X_test_processed)

# Probability predictions
y_prob_rf = rf_model.predict_proba(X_test_processed)[:, 1]

# Evaluation metrics
accuracy_rf = accuracy_score(y_test, y_pred_rf)
precision_rf = precision_score(y_test, y_pred_rf)
recall_rf = recall_score(y_test, y_pred_rf)
f1_rf = f1_score(y_test, y_pred_rf)
roc_auc_rf = roc_auc_score(y_test, y_prob_rf)

# Confusion Matrix
cm_rf = confusion_matrix(y_test, y_pred_rf)

print("===== Random Forest Results =====")
print("Accuracy :", accuracy_rf)
print("Precision:", precision_rf)
print("Recall   :", recall_rf)
print("F1 Score :", f1_rf)
print("ROC-AUC  :", roc_auc_rf)

print("\nConfusion Matrix:")
print(cm_rf)

print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```

```python

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score
)
from xgboost import XGBClassifier

# Separate features and target
X = df.drop(columns=["UPJO"])
y = df["UPJO"]

# Use identical split for both models
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

def evaluate_model(X_train, X_test, y_train, y_test):
    categorical = X_train.select_dtypes(
        include=["object", "category", "bool"]
    ).columns.tolist()

    numerical = [
        col for col in X_train.columns
        if col not in categorical
    ]

    numeric_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="median"))
    ])

    categorical_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ])

    preprocessor = ColumnTransformer([
        ("num", numeric_pipe, numerical),
        ("cat", categorical_pipe, categorical)
    ])

    model = Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", XGBClassifier(
            n_estimators=200,
            max_depth=4,
            learning_rate=0.05,
            subsample=0.9,
            colsample_bytree=0.9,
            eval_metric="logloss",
            random_state=42
        ))
    ])

    model.fit(X_train, y_train)

    predictions = model.predict(X_test)
    probabilities = model.predict_proba(X_test)[:, 1]

    return {
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(
            y_test, predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_test, predictions, zero_division=0
        ),
        "F1-score": f1_score(
            y_test, predictions, zero_division=0
        ),
        "ROC-AUC": roc_auc_score(y_test, probabilities)
    }

# Model A: All original features
results_all = evaluate_model(
    X_train, X_test, y_train, y_test
)

# Model B: Remove Surgical_modality
X_train_no_surgery = X_train.drop(
    columns=["Surgical_modality"]
)
X_test_no_surgery = X_test.drop(
    columns=["Surgical_modality"]
)

results_no_surgery = evaluate_model(
    X_train_no_surgery,
    X_test_no_surgery,
    y_train,
    y_test
)

comparison = pd.DataFrame({
    "With all features": results_all,
    "Without Surgical_modality": results_no_surgery
})

print(comparison.round(4))
print("\nDifference (without surgery - all features):")
print(
    (comparison["Without Surgical_modality"]
     - comparison["With all features"]).round(4)
)
```

```python

# Preoperative prediction sensitivity analysis

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score
)
from xgboost import XGBClassifier
import pandas as pd

# Exclude ID and information unavailable at preoperative prediction time
excluded = [
    "UPJO",
    "Patient_ID",
    "Postoperative_DRF_percent",
    "Surgical_modality"
]

X_preop = df.drop(columns=excluded)
y = df["UPJO"]

# Same split for both experiments
X_train, X_test, y_train, y_test = train_test_split(
    X_preop, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

def run_xgb(Xtr, Xte, ytr, yte):
    cat_cols = Xtr.select_dtypes(
        include=["object", "category", "bool"]
    ).columns.tolist()

    num_cols = [
        c for c in Xtr.columns if c not in cat_cols
    ]

    prep = ColumnTransformer([
        ("num", SimpleImputer(strategy="median"), num_cols),
        ("cat", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("onehot", OneHotEncoder(handle_unknown="ignore"))
        ]), cat_cols)
    ])

    pipe = Pipeline([
        ("prep", prep),
        ("model", XGBClassifier(
            n_estimators=200,
            max_depth=4,
            learning_rate=0.05,
            subsample=0.9,
            colsample_bytree=0.9,
            eval_metric="logloss",
            random_state=42
        ))
    ])

    pipe.fit(Xtr, ytr)
    pred = pipe.predict(Xte)
    prob = pipe.predict_proba(Xte)[:, 1]

    return {
        "Accuracy": accuracy_score(yte, pred),
        "Precision": precision_score(yte, pred, zero_division=0),
        "Recall": recall_score(yte, pred, zero_division=0),
        "F1-score": f1_score(yte, pred, zero_division=0),
        "ROC-AUC": roc_auc_score(yte, prob)
    }

# Model A: All available preoperative features
result_with_drf = run_xgb(
    X_train, X_test, y_train, y_test
)

# Model B: Remove preoperative DRF
X_train_no_drf = X_train.drop(
    columns=["Preoperative_DRF_percent"]
)
X_test_no_drf = X_test.drop(
    columns=["Preoperative_DRF_percent"]
)

result_without_drf = run_xgb(
    X_train_no_drf, X_test_no_drf, y_train, y_test
)

comparison = pd.DataFrame({
    "Preop features with DRF": result_with_drf,
    "Preop features without DRF": result_without_drf
})

print(comparison.round(4))
```
